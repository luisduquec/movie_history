# Preparación para carga incremental

Elimina y recrea los esquemas `movie_silver` y `movie_gold` para iniciar la carga incremental desde cero.

> ⚠️ `DROP SCHEMA ... CASCADE` borra todas las tablas de ambos esquemas.

### Reiniciar Silver

```sql
DROP SCHEMA IF EXISTS movie_silver CASCADE;

CREATE SCHEMA IF NOT EXISTS movie_silver
MANAGED LOCATION "abfss://silver@moviehistoryproject.dfs.core.windows.net"
```

### Reiniciar Gold

```sql
DROP SCHEMA IF EXISTS movie_gold CASCADE;

CREATE SCHEMA IF NOT EXISTS movie_gold
MANAGED LOCATION "abfss://gold@moviehistoryproject.dfs.core.windows.net"
```
