Recorre las carpetas de fecha de `bronze/incremental_append` y, para cada lote, ejecuta los notebooks de ingesta enviando `p_environment` y `p_file_date`.

### Lotes disponibles

Lista las carpetas de fecha en `bronze/incremental_append`.

```python
file_date_paths = dbutils.fs.ls("abfss://bronze@moviehistoryproject.dfs.core.windows.net/incremental_append")
display(file_date_paths)
```

### Ejecución por lote

Para cada fecha, ejecuta los notebooks cuyo nombre contiene `_file_` o `_folder_`.

```python
for folder in file_date_paths:
    for nb in ingestion_nb:
        if any(x in nb.name for x in ["_file_", "_folder_"]):
            file_date = folder.name[:-1]
            dbutils.notebook.run(nb.name, 0, {"p_environment": "dev", "p_file_date": file_date})
```
