# Configuración de accesos

Registra en Unity Catalog las ubicaciones externas de Bronze, Silver y Gold con la credencial `storage_credential_mov_his` y prueba el acceso al contenedor Bronze.

### Ubicaciones externas

Una ubicación por contenedor, todas asociadas a la misma credencial.

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS external_location_bronze_mov_his URL
'abfss://bronze@moviehistoryproject.dfs.core.windows.net'
WITH (STORAGE CREDENTIAL storage_credential_mov_his)
COMMENT 'External location for bronze data lakehouse'
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS external_location_silver_mov_his URL
'abfss://silver@moviehistoryproject.dfs.core.windows.net'
WITH (STORAGE CREDENTIAL storage_credential_mov_his)
COMMENT 'External location for silver data lakehouse'
```

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS external_location_gold_mov_his URL
'abfss://gold@moviehistoryproject.dfs.core.windows.net'
WITH (STORAGE CREDENTIAL storage_credential_mov_his)
COMMENT 'External location for gold data lakehouse'
```
