### Esquema Silver

Crea `movie_silver` con ubicación administrada en el contenedor Silver y muestra su descripción.

```sql
CREATE SCHEMA IF NOT EXISTS movie_silver
MANAGED LOCATION "abfss://silver@moviehistoryproject.dfs.core.windows.net"
```

```sql
DESCRIBE SCHEMA movie_silver
```

### Registro de tablas Silver

Convierte cada carpeta Parquet en una tabla Delta. Las primeras celdas usan PySpark (`saveAsTable`) y las siguientes SQL (`CREATE OR REPLACE TABLE`); el resultado es equivalente.

```sql
%python
movie_df_silver = (
    spark.read
        .format("parquet")
        .load(f"{silver_folder_path}/movies")
)

(
    movie_df_silver.write
        .format("delta")
        .mode("overwrite")
        #.partitionBy("year_release_date")
        .saveAsTable("movie_silver.movies")
    
)
```
