```sql
CREATE SCHEMA IF NOT EXISTS movie_bronze
```

```sql
DROP TABLE IF EXISTS movie_bronze.movie;
CREATE TABLE movie_bronze.movie
AS
SELECT *
FROM read_files(
    'abfss://bronze@moviehistoryproject.dfs.core.windows.net/movie.csv',
    format => 'csv',
    header => true,
    schema => 'movie_id INT, title STRING, budget INT, homePage STRING, overview STRING, popularity FLOAT, yearReleaseDate INT, releaseDate DATE, revenue INT, durationTime INT, movieStatus STRING, tagline STRING, voteAverage FLOAT, voteCount INT'
);
```
