```sql
SELECT
    country_name, 
    COUNT(country_name) AS count_movies,
    SUM(budget) AS total_budget,
    CAST(AVG(budget) AS INTEGER) AS avg_budget,
    SUM(revenue) AS total_revenue,
    CAST(AVG(revenue) AS INTEGER) AS avg_revenue
FROM results_movie
GROUP BY country_name
ORDER BY total_revenue DESC
```

```sql
CREATE OR REPLACE TEMP VIEW WINDOW_MOVIE AS
SELECT
    country_name, 
    year_release_date,
    SUM(budget) AS total_budget,
    SUM(revenue) AS total_revenue
FROM results_movie
WHERE year_release_date IS NOT NULL
GROUP BY country_name, year_release_date
ORDER BY country_name, year_release_date ASC
```

```sql
SELECT * FROM WINDOW_MOVIE
```

```sql
SELECT
    country_name,
    year_release_date,
    total_budget,
    sum(total_budget) OVER(PARTITION BY country_name ORDER BY year_release_date ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS cumulative_budget,
    total_revenue,
    sum(total_revenue) OVER(PARTITION BY country_name ORDER BY year_release_date ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS cumulative_revenue
FROM WINDOW_MOVIE
ORDER BY country_name ASC, year_release_date ASC
```
