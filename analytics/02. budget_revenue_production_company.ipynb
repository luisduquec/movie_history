### Métricas por compañía productora

Calcula el número de películas y las métricas de presupuesto e ingresos para cada compañía.

```sql
SELECT
    company_name, 
    COUNT(country_name) AS count_movies,
    SUM(budget) AS total_budget,
    CAST(AVG(budget) AS INTEGER) AS avg_budget,
    SUM(revenue) AS total_revenue,
    CAST(AVG(revenue) AS INTEGER) AS avg_revenue
FROM results_movie
GROUP BY company_name
ORDER BY total_revenue DESC
```
