### Ejecución por lote

Para cada fecha, ejecuta los notebooks cuyo nombre no empieza por `__`.

```python
for folder in file_date_paths:
    for notebook in [nb for nb in transform_nb if not(nb.name.startswith("__"))]:
        file_date = folder.name[:-1]
        dbutils.notebook.run(notebook.name, 0, {"p_file_date": file_date})
```
