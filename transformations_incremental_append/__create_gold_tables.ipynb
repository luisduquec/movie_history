```sql
%sql
CREATE SCHEMA IF NOT EXISTS movie_gold
MANAGED LOCATION 'abfss://gold@moviehistoryproject.dfs.core.windows.net'
```

```sql
CREATE OR REPLACE TABLE movie_gold.results_movie_genre_language
USING DELTA
AS
SELECT *
FROM
parquet.`abfss://gold@moviehistoryproject.dfs.core.windows.net/results_movie_genre_language`
```

```sql
CREATE OR REPLACE TABLE movie_gold.results_country_prod_company
USING DELTA
AS
SELECT *
FROM
parquet.`abfss://gold@moviehistoryproject.dfs.core.windows.net/results_country_prod_company`
```

```sql
CREATE OR REPLACE TABLE movie_gold.results_group_movie_genre
USING DELTA
AS
SELECT *
FROM
read_files('abfss://gold@moviehistoryproject.dfs.core.windows.net/results_movie_genre_country',
format => 'parquet'
)
```

```sql
CREATE OR REPLACE TABLE movie_gold.results_group_movie_country
USING DELTA
AS
SELECT *
FROM
read_files(
    'abfss://gold@moviehistoryproject.dfs.core.windows.net/results_group_movie_country',
    format => 'parquet'
);
```
