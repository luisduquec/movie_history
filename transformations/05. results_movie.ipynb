### Construcción de results_movie

Une películas con países y compañías productoras y guarda año, país, compañía, presupuesto e ingresos.

```sql
CREATE OR REPLACE TABLE movie_gold.results_movie
USING DELTA
AS
SELECT M.year_release_date, C.country_name, PCO.company_name, M.budget, M.revenue
FROM movies as M
INNER JOIN production_countries as PC ON M.movie_id = PC.movie_id
INNER JOIN countries AS C ON PC.country_id = C.country_id
INNER JOIN movies_companies AS MC ON M.movie_id = MC.movie_id
INNER JOIN production_companies AS PCO ON MC.company_id = PCO.company_id
```
